## Tarea 2 · Python como calculadora

```python
total_tiendas = 120
tickets_por_tienda = 1000
lineas_por_ticket = 8
print(f"Colmena Market general al día {total_tiendas * tickets_por_tienda * lineas_por_ticket} líneas de ticket")
```

## Tarea 3 · Las cámaras de frío, con variables

```python
camaras_por_tienda = 3
lecturas_por_hora = 12
horas_dia = 24
dias_año = 365
lecturas_al_dia = total_tiendas * camaras_por_tienda * lecturas_por_hora * horas_dia
print(f"Lecturas al día: {lecturas_al_dia}")
print(f"Lecturas al año: {lecturas_al_dia * dias_año}")
```

## Tarea 4 · De lecturas a megas

```python
bytes_por_lectura = 46
megas_al_dia = total_tiendas * camaras_por_tienda * lecturas_por_hora * horas_dia* bytes_por_lectura / 1000000
gigas_al_dia = megas_al_dia * dias_año / 1000
print(f"Sensores: {megas_al_dia:.2f} MB al día")
print(f"Sensores: {gigas_al_dia:.2f} GB al año")

```
